# Downloading 100 files

```python
count = 0

for line in text:
  Attributes = line.split()

  if len(line.split()) >= 5 and line.split()[-4] == "10-K":
    URL = websitePrefix + Attributes[-1]
    fileDate = Attributes[-2]
    CIK = Attributes[-3]
    formType = Attributes[-4]
    formType.replace('/', '_')

    response = requests.get(URL, headers=headers)
    count += 1

    fileName = str(count) + "_" + formType + "_" + CIK + "_" + fileDate + ".txt"

    with open(downloadDir + fileName, "w") as f:
      f.write(response.text)

    print(count, fileName)

  if count == 100:
    break

print("Downloaded", count, "files")
```

```python
import os
import re

word_counts = {}

for fileName in os.listdir(downloadDir):
  if fileName.endswith(".txt"):
    with open(downloadDir + fileName, "r") as f:
      content = f.read().lower()

    words = re.findall(r"[a-z]+", content)

    for word in words:
      if word in word_counts:
        word_counts[word] = word_counts[word] + 1
      else:
        word_counts[word] = 1
```
