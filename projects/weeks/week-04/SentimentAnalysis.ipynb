# Positive and Litigious Words Count

```python
import pandas as pd

lm = pd.read_csv("/content/drive/MyDrive/Fall 2026 Text Mining/week 03/Lanqing Jia/Loughran-McDonald_MasterDictionary_1993-2025.csv")
print(lm.head())
```

```python
lm["Word"] = lm["Word"].astype(str).str.lower()

positive  = set(lm.loc[lm["Positive"]  != 0, "Word"])
negative  = set(lm.loc[lm["Negative"]  != 0, "Word"])
litigious = set(lm.loc[lm["Litigious"] != 0, "Word"])

print(len(positive), len(negative), len(litigious))
```

```python
import re, os, glob

DOCS = "/content/drive/MyDrive/Fall2026TextMining_me/Lanqing10Ks"
files = sorted(glob.glob(os.path.join(DOCS, "*.txt")))
len(files)
```

```python
rows = []
pos, neg, lit = 0, 0, 0

for path in files:
  text = open(path, "r", encoding="utf-8", errors="ignore").read().lower()
  words = re.findall(r"[a-z']+", text)
  n = len(words)

  for w in words:
    if w in positive:
      pos += 1
    elif w in negative:
      neg += 1
    elif w in litigious:
      lit += 1

  rows.append([os.path.basename(path), n, pos, neg, lit, pos / n, neg / n, lit / n])

df = pd.DataFrame(rows, columns=["fileName", "wordCount", "PosWordCount", "NegWordCount", "LitWordCount", "PosRate", "NegRate", "LitRate"])
len(df)
```
