## Primitive Data Types
<a id="primitive-data-types"></a>

Python has a small set of **primitive** (fundamental) types:

| Type | Description | Literal Examples |
|---|---|---|
| `int` | Whole numbers | `0`, `-3`, `42` |
| `float` | Decimal numbers (double precision) | `3.14`, `-0.5`, `1.0` |
| `bool` | Boolean values | `True`, `False` |
| `str` | Text (Unicode strings) | `"hello"`, `'world'` |
| `NoneType` | The absence of a value | `None` |
| `complex` | Complex numbers *(optional)* | `3+4j`, `-2j` |

**Tip:** Use `type(x)` to see a value's type.


```python
# Try changing these values and re-run
a = 42           # int
b = 3.14         # float
c = True         # bool
d = "hello"      # str
e = None         # NoneType
z = 2 + 3j       # complex

print("a", "b", "f")
print("Lanqing")
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
print(e, type(e))
print(z, type(z))

```


## Numbers & Operators
<a id="numbers-operators"></a>

Common numeric operators:

- `+` add, `-` subtract, `*` multiply, `/` divide → **float**
- `//` floor division (truncates toward negative infinity)
- `%` modulo (remainder), `**` exponent
- Comparisons: `==`, `!=`, `<`, `<=`, `>`, `>=`


```python
x, y = 7, 3
print("x + y =", x + y)
print("x - y =", x - y)
print("x * y =", x * y)
print("x / y =", x / y)     # float
print("x // y =", x // y)   # floor division
print("x % y =", x % y)     # modulo
print("x ** y =", x ** y)   # exponent

print("comparisons:", x == y, x != y, x > y, x <= y)
```


## Strings & Slicing
<a id="strings-slicing"></a>

A string is a sequence of characters. **Slicing** lets you grab parts of sequences (strings, lists, tuples):

- `seq[start:stop]` → from `start` to `stop-1`
- `seq[start:stop:step]` → jump by `step`
- Omit any part to use defaults (`start=0`, `stop=len(seq)`, `step=1`)
- Negative indices count from the end.

Try these and experiment!


```python
text = "Hello, World!"
print("Original:", text)
print("text[0:5]  ->", text[0:5])    # 'Hello'
print("text[7:]   ->", text[7:])     # 'World!'
print("text[:5]   ->", text[:5])     # 'Hello'
print("text[::2]  ->", text[::2])    # 'Hlo ol!'
print("text[::-1] ->", text[::-1])   # reverse string

alist = [1,2,3,4,5]
print(alist[-2:])

```

```python
# Slicing lists works the same way
nums = list(range(10))
print("nums:", nums)
print("nums[2:8]  ->", nums[2:8])
print("nums[::3]  ->", nums[::3])
print("nums[1:9:2]->", nums[1:9:2])
```


## Control Flow
<a id="control-flow"></a>

Use **conditionals** and **loops** to control the execution of your program.


### If / Elif / Else

```python
score = 50
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D/F"

print(f"score={score} -> grade={grade}")
```

### For Loops (with `range` and `enumerate`)

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)

print("Using range:")
print (len(fruits))
print (range(len(fruits)))
for i in range(len(fruits)):
    print(i)
    print(i, fruits[i])

```

```python
print("\nUsing enumerate:")
print (enumerate(fruits))
for item in enumerate(fruits, start=1):
    print(item)

```

```python
# Slicing inside a loop
print("\nEvery other fruit via slicing:")
for item in fruits[::2]:
    print(item)
```

### While Loops, `break`, `continue`, and Loop `else`

```python

n = 0
total = 0
while True:
    n += 1
    if n % 2 == 1:   # skip odd numbers
        continue
    total += n
    print(total)
    if total >= 20:
        break
else:
    # Runs only if the while loop didn't break
    print("Loop finished normally.")
print("total:", total)
```

### `any` and `all` (short‑circuit logic)

```python

checks = [
    lambda s: len(s) >= 8,
    lambda s: any(c.isupper() for c in s),
    lambda s: any(c.islower() for c in s),
    lambda s: any(c.isdigit() for c in s),
]

def strong_password(pw):
    return all(check(pw) for check in checks)

for pw in ["Abc123", "Abcdef12", "abcDEF12"]:
    print(pw, "->", strong_password(pw))
```


## Comprehensions
<a id="comprehensions"></a>

Comprehensions create new collections concisely.

- **List:** `[expr for x in iterable if condition]`
- **Set:** `{expr for x in iterable}`
- **Dict:** `{key_expr: val_expr for x in iterable}`
- **Generator:** `(expr for x in iterable)` *(lazy)*


```python

nums = [1, 2, 3, 4, 5, 6]
# List comprehension: squares of even numbers
squares_even = [x*x for x in nums if x % 2 == 0]
print("list comp:", squares_even)

```

```python
# Set comprehension: lengths of words (unique)
words = ["apple", "banana", "pear", "banana"]
lengths = {len(w) for w in words}
print("set comp:", lengths)

```

```python
# Dict comprehension: map word -> length
word_len = {w: len(w) for w in words}
print("dict comp:", word_len)

```

```python
# Nested list comp: flatten a matrix
matrix = [[1,2,3], [4,5], [6,7,8]]
flattened = [x for row in matrix for x in row]
print("flattened:", flattened)

```

```python
# Gentle generator example (use next() or iterate)
gen = (x*x for x in range(3))
print("generator next:", next(gen), next(gen), next(gen))
```


## Functions & Lambdas
<a id="functions-lambdas"></a>

Define reusable blocks with `def`. Use **guard clauses** to fail fast. **Lambdas** are tiny anonymous functions.


```python

def safe_divide(a, b=1):
    """Divide a by b; return a message on division by zero."""
    if b == 0:               # guard clause
        return "undefined (division by zero)"
    return a / b

print("10 / 2 =", safe_divide(10, 2))
print("10 / 0 =", safe_divide(10, 0))

```


## Exceptions & Error Handling
<a id="exceptions-error-handling"></a>

Use `try / except / else / finally` to handle errors gracefully. Use `raise` to signal errors.


```python

def to_int(s):
    try:
        value = int(s)
    except ValueError:
        print(f"Cannot convert {s!r} to int.")
        return None
    else:
        # runs only if no exception
        return value
    finally:
        print("Attempted conversion.")

print("OK  :", to_int("123"))
print("Fail:", to_int("12.3"))

```

```python
def sqrt_nonneg(x):
    if x < 0:
        raise ValueError("x must be non‑negative")
    return x ** 0.5

try:
    print(sqrt_nonneg(9))
    print(sqrt_nonneg(-1))
except ValueError as err:
    print("Caught error:", err)
```


## Example: Gradebook Analyzer
<a id="mini-exercise-gradebook-analyzer"></a>

*Goal:* Given a list of scores, print each student's **letter grade**, class **average**, and the **top 3** scores.  
This pulls together numbers, lists, slicing, conditionals, sorting, and comprehensions.


```python

scores = [87, 92, 76, 100, 68, 84, 91, 73, 88]

def letter(n):
    if   n >= 90: return "A"
    elif n >= 80: return "B"
    elif n >= 70: return "C"
    elif n >= 60: return "D"
    else:         return "F"

# Print each score with its letter
for i, s in enumerate(scores, start=1):
    print(f"Student {i}: {s} -> {letter(s)}")

# Average (float)
avg = sum(scores) / len(scores)

# Top 3 using slicing after sorting (descending)
top3 = sorted(scores, reverse=True)[:3]
print("\nAverage:", round(avg, 2))
print("Top 3  :", top3)

# Bonus: dictionary mapping score -> letter using a dict comprehension
grade_map = {s: letter(s) for s in scores}
print("Grade map:", grade_map)
```
